2. User-Defined Functions

we define function by using the (def) keyword.

```python
# Basic function
def greet():
    print("Hello, Students")

greet()    
```

```python
# Function with arguments and return statement
def add_numbers(a=6,b=0):
    """This function adds two numbers and returns the result.
    
    Default value 1, 0 respectively"""
    return a+b
```

```python
result =add_numbers(20,-9)
print(f"The sum is:{result}")
```

```python
# this give 
print(add_numbers.__doc__)
```

```python

# Function with Parameters

# Function accepts input values

def welcome(name):
    print("Hello",name)
welcome("mohit")
welcome("Harsh")
```

```python
# Addition Function

def add(a,b):
    result=a+b
    return result
answer=add(9,9)
print(answer)
```

```python
# Square of Number
def square(number):
    return number*number
print(square(6))
```

```python
# Even or Odd Checker
def check_even(number):
    if number%2==0:
        return "Even"
    else:
        return "odd"

print(check_even(9))
print(check_even(0))
```

```python
# Real-Life Examples

# Student Grade 

def grade (marks):
    if marks>=90:
        return "Grade A"
    elif marks>=75:
        return "Grade B"
    else:
        return "Fail"

print(grade(89))    
print(grade(9))    
print(grade(90))    
```

3. Lambda Functions

Lambdas are small anonymous functions defined with the lambda keyword.

```python
def square(x):
    sq_result = x*x
    return sq_result
```

```python
square = lambda x: x*x
```

```python
# Lambda function to square a number
square =lambda x:x ** 2

print(f"square of 5-> {square(4)}")

# Lambda with multiple arguments
multiply= lambda x,y: x*y
print(f"Product of 10 and 4-> {multiply(10,4)} ")
```

```python
# Normal function
def add(a,b):
    return a+b
print(add(10,2))

# Lambda version
add_lambda = lambda a,b:a+b
print(add_lambda(10,20))
```

```python
## Find Largest of Three Numbers
largest =lambda a,b,c:a if a>b and a>c else (b if b>c else c)
print(largest(25,50,40))
```

```python
# Lambda with map()
numbers=[1,2,3,4,5]
squares=list(map(lambda x:x*x,numbers))
print(squares)
```

```python
# Convert Names to Uppercase

names=["mohit","rahul","ankit"]

upper_Case=list(map(lambda x:x.upper(),names))

print(upper_Case)
```

```python
# Filter Students with Marks > 75

marks=[45,88,92,60,77,95]

top_students=list(filter(lambda x:x>75,marks))
print(top_students)
```

1. LIST COMPREHENSION

```python
# List comprehension provides a shorter syntax
# to create lists.

# Traditional method

numbers=[]

for i in range(1,11):
    numbers.append(i)
    
print(numbers)  
```

```python
# List comprehension version
numbers=[x for x in range(1,11)]
print(numbers)  
```

Explanation:
Read from right to left:
for every i in range(1,11)
put i into list

```python
# ============================================================
# EXAMPLE 2: SQUARE OF NUMBERS
# ============================================================

squares=[x*x for x in range(1,11)]
print(squares)  

#============================================================
# EXAMPLE 3: EVEN NUMBERS
# ============================================================

even=[x for x in range(1,11) if x%2==0]
print(even)
```

```python
# ============================================================
# EXAMPLE 4: CONVERT NAMES TO UPPERCASE
# ============================================================

students=["mohit","sachin","rahul","rohit"]
upper_names=[name.upper() for name in students]
print(upper_names)
```

```python
# ============================================================
# REAL LIFE EXAMPLE
# Student marks greater than 75
# ============================================================

marks=[80, 90, 70, 85, 95, 75, 88, 92]
top_students=[x for x in marks if x>75]
print(top_students)
```

```python
#============================================================
#MINI TASK FOR STUDENTS
#============================================================
#Create list of cubes from 1–10

cubes=[x**3 for x in range(1,11)]
print(cubes)
```

2. DICTIONARY COMPREHENSION

```python
# Dictionary comprehension creates
# dictionaries quickly

# Syntax:
# {key:value for item in iterable}

numbers=[1,2,3,4,5]
square_dict={x:x*x for x in numbers}
print(square_dict)
```

```python
# ============================================================
# EXAMPLE 3
# Filter even values
# ============================================================

even_square={x:x*x for x in range(1,11) if x%2==0}
print(even_square)
```

```python
# ============================================================
# REAL LIFE EXAMPLE
# Student Grades
# ============================================================

students=["mohit","sachin","rahul","rohit"]
grades=["A","B","C","A"]
students_grade={students[i]:grades[i] for i in range(len(students))}

print(students_grade)
```

```python
# ============================================================
# CREATE/WRITE FILE
# ============================================================

file=open("students.txt","w")
file.write("Welcome Students\n")
file.write("praticing file handling\n")
file.close()
print("File created and data written successfully.")
```

```python
# ============================================================
# READ FILE
# ============================================================

file=open("students.txt","r")
content=file.read()
print("file content:\n",content)
file.close()
```

```python
# ============================================================
# APPEND DATA
# ============================================================

file=open("students.txt","a")
file.write("learning file handling \n")
file.close()
print("New content added ")
```

```python
# ============================================================
# READ LINE BY LINE
# ============================================================

file=open("students.txt","r")
for line in file:
    print(line)

file.close()
```

```python
# ============================================================
# BEST PRACTICE: WITH
# ============================================================

# with automatically closes file

with open("students.txt","r") as file:
    content=file.read()
    print("file content:\n",content)
```

```python
# ============================================================
# REAL LIFE EXAMPLE
# STUDENT REPORT
# ============================================================

with open("report.txt","w") as file:
    file.write("Student Report\n")
    file.write("Name: Mohit\n")
    file.write("Marks: 85\n")
    file.write("Grade: A\n")
print("reprort.txt file created successfully.")    
```

EXCEPTION HANDLING

```python
#Exception:
# Runtime error during execution
```

```python
print("\nEXCEPTION HANDLING")
```

```python
# ============================================================
# BASIC TRY EXCEPT
# ============================================================

try:
    number=10/0
except:
    print("Error occurred")    
print("Program continues...")    
```

```python
# ============================================================
# SPECIFIC EXCEPTION
# ============================================================

try:
    num=int(input("Enter Number: "))
except ValueError:
    print("Please enter a valid integer.")    
```

```python
# ============================================================
# MULTIPLE EXCEPT
# ============================================================

try:
    number=int(input("Enter a number: "))
    result=10/number
    print("Result:",result)
except ValueError:
    print("Please enter a valid integer.")
except ZeroDivisionError:
    print("Division by zero is not allowed.")        
```

```python
# ============================================================
# ELSE BLOCK
# ============================================================

try:
    a=10
    b=2
    result=a/b
    print("Result:", result)
except:
    print("Error")
else:
    print("Division Success")        
```

```python
# ============================================================
# FINALLY BLOCK
# ============================================================

try:
    file=open("demo.txt")
except:
    print("File not found")    
finally:
    print("Execution completed")    
```

```python
# FileNotFoundError

try:
    open("abc.txt")
except FileNotFoundError:
    print("File does not exist")    
```

```python
# ============================================================
# USER DEFINED EXCEPTION
# ============================================================

try:
    age=15
    if age<18:
        raise Exception("Age must be 18 or older.")
except Exception as e:
        print("Error:", e)
```

# List in python

List stores multiple values and is mutable.

```python
# Creating a list 
students = ["John", "Mike", "Sarah", "Jessica"]
print(students)  

#access  by index
print(students[0]) 
print(students[-4]) 

#Add elemts
students.append("David")
print(students)

#Rmove elements
students.remove("Mike")
print(students)

del students[1]
print(students)
```

**Tuple in python**

Tuples is immutable (cannot be changed)

```python
colors=("Red", "Green", "Blue", "Yellow")

print(colors)
print(colors[0])

numbers = (1, 2, 3, 4, 5)
print(numbers.count(3))  
print(numbers.index(4))
```

**Dictionary in python**

```python
students={
    "name": "John",
    "age": 20,
    "course": "python"
}

print(students)
print(students["name"])

students["city"]="Delhi"
print(students)
print(students.keys())
print(students.values())
```

**set in python**

set remove duplicate values

```python
marks ={90, 80, 70, 60, 50}
print(marks)

marks.add(95)
print(marks)

marks.remove(70)
print(marks)
```

**Looping with Data Structures**

```python
students=["John", "Mike", "Sarah", "Jessica"]

for students in students:
    print(students)
```

**Break Continue Pass**

```python
#Break -->stop

for i in range (1, 10):
    if i == 5:
        break
    print(i)
print("******")

#continue -->skip    

for i in range(1, 8):
    if i == 5:
        continue
    print(i)
print("******")
#Pass

for i in range(1, 5):
    if i == 3:
        pass
print("Pass executed")    
```

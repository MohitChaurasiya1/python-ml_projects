1. Operators in Python

```python
# Arithmetic operators
 
a= -40
b= 4

print("Addition = ", a + b)
print("Subtraction = ", a - b)
print("Multiplication = ", a * b)
print("Division = ", a / b)
print("Modulus = ", a % b)
print("Floor Division = ", a // b)
print("Power = ", a ** b)
```

Comparison Operators

```python
x= 20
y=15

print("x == y : ", x == y)
print("x != y : ", x != y)
print("x > y : ", x > y)
print("x < y : ", x < y)
print("x >= y : ", x >= y)
print("x <= y : ", x <= y)
```

Assignments Operators

```python
num = num + 5

num += 5
```

```python
num = 10
print("Initial value = ", num)

num += 5
print("After addition = ", num)

num -= 5
print("After subtraction = ", num)

num *= 5
print("After multiplication = ", num)

num /= 5
print("After division = ", num)
```

Logical Operators

```python
age = 20
salary = 50000

print(age > 18 and salary > 30000)  # True
print(age > 18 or salary > 30000)   # True
print(not (age > 18))                # False
```

Membership Operators

```python
students_list = ["Alice", "Bob", "Charlie", "David" , "Mohit"]

print("Mohit is in the list: ", "Mohit" in students_list)  # True
print("Aman is in the list: ", "Aman" not in students_list)    # False
```

Identity Operators

```python
a= 10
b= 20

print(a is b)  # False
```

## 2. Control Statements(Conditions)


```python
# simple if Condition
age= 20
if age >= 18:
    print("You are eligible to vote.")
else:
    print("You are not eligible to vote.")    
```

```python
# simple if-else Condition
num= 10
if num % 2 == 0:
    print("The number is even.")
else: 
    print("The number is odd.")    
```

```python
# simple if-elif-else Condition

marks= 55

if marks >= 90:
    print("Grade: A")
elif marks >= 80:
    print("Grade: B")
elif marks >= 70:
    print("Grade: C")
else:
    print("fail")            
```

# Nested if statements

```python
username= "admin"
password= "admin123"

if username == "admin":

    if password == "admin123":
        print("Login successful!")
    else:
        print("Incorrect password.")
else:
    print("Incorrect username.")        
```

Positive, Negative or Zero Program 

```python
num = 0
if num > 0:
    print("The number is positive.")
elif num < 0:
    print("The number is negative.")
else:
    print("The number is zero.")        
```

**2. NumPy Array vs Python List**

Size: NumPy arrays take up significantly less memory in your machine.
Performance: They are massively faster. NumPy is largely written in C and optimizes memory allocation.
Functionality: NumPy arrays have built-in, highly optimized functions for complex mathematical operations.

**3. Creation of 1D, 2D and 3D array**

The core of NumPy is the ndarray (N-dimensional array). An array's dimensionality is known as its rank.

1D Array: A list of values (Vector)
2D Array: A list of lists (Matrix / Table with rows and columns)
3D Array: A list of lists of lists (Cube / Tensor)

```python
# 1. Creating a 1d Array
array_1d =np.array([10,20,30,40,50])
print('---1d Array---')
print(array_1d)
print('Dimensions:',array_1d.ndim)
print('Shape:',array_1d.shape) # output is (5,) meaning 5 items in 1 dimension
```

```python
# 2. Creating a 2D Array (Pass a list of lists)
array_2d=np.array([[10,20,30],[40,50,60],[70,80,90]])

print('---2d Array---')
print(array_2d)
print('Dimensions:',array_2d.ndim)
print('Shape:',array_2d.shape) # output is (3,3) meaning 3 rows and 3 columns
```

```python
# 3. Creating a 3D Array (Pass a list of lists of lists)
array_3d = np.array([[[1, 2], [3, 4]], [[5, 6], [7, 8]]])

print('---3d Array---')
print(array_3d)
print('Dimensions:',array_3d.ndim)
print('Shape:',array_3d.shape) # output is (2, 2,
```

**4 & 5. Special NumPy Functions**

Often, we don't type out arrays by hand. We need to generate arrays of a specific shape to hold data later. NumPy provides powerful built-in functions for this:

np.zeros(): Fills array with 0s.
np.ones(): Fills array with 1s.
np.full(): Fills array with a specific number.
np.arange(): Creates sequences of numbers.

```python
# np.zeros() - Great for initializing placeholders
zeros_matrix = np.zeros((3, 4)) # 3 rows, 4 columns
print('---Zeros Matrix---')
print(zeros_matrix)
```

```python
# np.ones() 
ones_matrix = np.ones((2, 2))
print('---Ones Matrix---')
print(ones_matrix)
```

```python
# np.full() - Specify shape and the value you want to fill it with
full_matrix = np.full((3, 3), 99) 
print('---Full Matrix---')
print(full_matrix)
```

```python
# np.arange() - Start, Stop (exclusive), Step
sequence_array = np.arange(0, 20, 2)
print('---Sequence Array---')
print(sequence_array)
```

**Hands On: N-dimensional Arrays and Operations**
Let's put it all together. Operations in NumPy are vectorized, meaning they apply to the entire array at once without needing a for loop.

```python
# Setup a 2D Array representing test scores for 3 students across 4 subjects
scores = np.array([[85, 90, 88, 92],
                   [78, 80, 85, 88],
                   [92, 95, 98, 100]])

print('Original Scores Array:')
print(scores)
print('-'*40)
```

```python
# 1. Scalar Operations (Curve everyone's grade by +5 points)
curved_scores = scores + 5
print('Curved Scores Array:')
print(curved_scores)
```

```python
# 2. Slicing (Extract just Student 1's scores)
# Syntax: array[row_start:row_end, col_start:col_end]
student_1 = scores[0, :]
print('Student 1 Scores:', student_1)
```

```python
# Extract just Subject 3's scores for all students
subject_3 = scores[:, 2]
print('Subject 3 Scores:', subject_3)
print('-'*40)
```

```python
# 3. Statistical Operations
print('Maximum score in the class:', np.max(scores))
print('Minimum score in the class:', np.min(scores))
print('Average score in the class:', np.mean(scores))
```

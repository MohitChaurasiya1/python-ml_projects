# 🐍 Python Variables

A comprehensive introduction to Python variables — types, naming rules, and operations.

---

## 1. What is a Variable?

A **variable** is a container for storing data values. In Python, you don't need to declare a variable type — Python infers it automatically.

```
variable_name = value
```

````python
# Creating your first variables
name = "Mohit"
age = 25
height = 5.9
is_student = True

print("Name:", name)
print("Age:", age)
print("Height:", height)
print("Is Student:", is_student)
````

## 2. Variable Types

Python has several built-in data types:

| Type | Example | Description |
|------|---------|-------------|
| `int` | `10` | Integer numbers |
| `float` | `3.14` | Decimal numbers |
| `str` | `"hello"` | Text strings |
| `bool` | `True` | Boolean values |
| `list` | `[1, 2, 3]` | Ordered collection |
| `tuple` | `(1, 2, 3)` | Immutable collection |
| `dict` | `{"a": 1}` | Key-value pairs |
| `set` | `{1, 2, 3}` | Unique elements |

````python
# Integer
x = 10
print(f"x = {x}, type: {type(x)}")

# Float
pi = 3.14159
print(f"pi = {pi}, type: {type(pi)}")

# String
greeting = "Hello, World!"
print(f"greeting = {greeting}, type: {type(greeting)}")

# Boolean
is_active = True
print(f"is_active = {is_active}, type: {type(is_active)}")
````

## 3. Variable Naming Rules

### ✅ Valid Names
- Must start with a letter or underscore (`_`)
- Can contain letters, numbers, and underscores
- Case-sensitive (`age` and `Age` are different)

### ❌ Invalid Names
- Cannot start with a number
- Cannot contain spaces or special characters
- Cannot use Python reserved keywords

````python
# ✅ Valid variable names
my_name = "Python"
_private = "hidden"
age2 = 30
firstName = "John"
MAX_VALUE = 100

print(my_name, _private, age2, firstName, MAX_VALUE)
````

````python
# ❌ These would cause errors (uncomment to test)
# 2name = "error"       # Cannot start with number
# my-name = "error"     # Cannot use hyphens
# my name = "error"     # Cannot use spaces
# class = "error"       # Cannot use reserved keywords
````

## 4. Multiple Assignment

Python allows you to assign values to multiple variables in a single line.

````python
# Multiple assignment
a, b, c = 1, 2, 3
print(f"a = {a}, b = {b}, c = {c}")

# Same value to multiple variables
x = y = z = 0
print(f"x = {x}, y = {y}, z = {z}")

# Swapping variables
a, b = b, a
print(f"After swap: a = {a}, b = {b}")
````

## 5. Type Conversion (Casting)

You can convert between data types using built-in functions.

````python
# String to Integer
num_str = "42"
num_int = int(num_str)
print(f"String '{num_str}' → Integer {num_int}, type: {type(num_int)}")

# Integer to Float
num_float = float(num_int)
print(f"Integer {num_int} → Float {num_float}, type: {type(num_float)}")

# Float to Integer (truncates decimal)
pi = 3.14159
pi_int = int(pi)
print(f"Float {pi} → Integer {pi_int}")

# Any type to String
age = 25
age_str = str(age)
print(f"Integer {age} → String '{age_str}', type: {type(age_str)}")
````

## 6. String Operations with Variables

````python
first_name = "John"
last_name = "Doe"
age = 28

# Concatenation
full_name = first_name + " " + last_name
print("Full Name:", full_name)

# f-strings (recommended)
intro = f"My name is {first_name} {last_name} and I am {age} years old."
print(intro)

# String methods
print(f"Upper: {full_name.upper()}")
print(f"Lower: {full_name.lower()}")
print(f"Length: {len(full_name)}")
print(f"Replace: {full_name.replace('John', 'Jane')}")
````

## 7. Arithmetic Operations

````python
a = 15
b = 4

print(f"a + b = {a + b}")       # Addition
print(f"a - b = {a - b}")       # Subtraction
print(f"a * b = {a * b}")       # Multiplication
print(f"a / b = {a / b}")       # Division (float)
print(f"a // b = {a // b}")     # Floor Division
print(f"a % b = {a % b}")       # Modulus (remainder)
print(f"a ** b = {a ** b}")     # Exponentiation
````

## 8. Collection Variables

````python
# List — ordered, mutable
fruits = ["apple", "banana", "cherry"]
print(f"Fruits: {fruits}")
print(f"First fruit: {fruits[0]}")

# Tuple — ordered, immutable
coordinates = (10.5, 20.3)
print(f"Coordinates: {coordinates}")

# Dictionary — key-value pairs
student = {
    "name": "Alice",
    "age": 22,
    "grade": "A"
}
print(f"Student: {student}")
print(f"Name: {student['name']}")

# Set — unique, unordered
unique_numbers = {1, 2, 3, 3, 4, 4, 5}
print(f"Unique Numbers: {unique_numbers}")
````

## 9. Dynamic Typing

Python is **dynamically typed** — a variable can change its type during execution.

````python
# Variable starts as integer
x = 10
print(f"x = {x}, type: {type(x)}")

# Now it's a string
x = "Hello"
print(f"x = {x}, type: {type(x)}")

# Now it's a list
x = [1, 2, 3]
print(f"x = {x}, type: {type(x)}")

# Now it's a boolean
x = True
print(f"x = {x}, type: {type(x)}")
````

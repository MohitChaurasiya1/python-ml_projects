```python
# Quest 3:- Write a program to check whether a student has passed. Conditions ● Marks ≥ 90 → Grade A ● Marks ≥ 75 → Grade B ● Marks ≥ 60 → Grade C ● Marks ≥ 40 → Grade D ● Otherwise Fail

marks = float(input("Enter marks: "))

print("Marks:", marks)

if marks >= 90:
    print("Grade A")
elif marks >= 75:
    print("Grade B")
elif marks >= 60:
    print("Grade C")
elif marks >= 40:
    print("Grade D")
else:
    print("Fail")
```

```python
# quest 6:- Create a dictionary of five students. Store ● Name ● Age ● Marks Perform ● Add new student ● Delete one student ● Update marks ● Print all students.

students = {
    "Student1": {"Name": "Rahul", "Age": 20, "Marks": 85},
    "Student2": {"Name": "Amit", "Age": 21, "Marks": 78},
    "Student3": {"Name": "Priya", "Age": 20, "Marks": 92},
    "Student4": {"Name": "Naveen", "Age": 21, "Marks": 88},
    "Student5": {"Name": "Rohit", "Age": 22, "Marks": 75}
}


print("Initial Dictionary:")
print(students)

# Add new student
students["Student6"] = {
    "Name": "Mohit",
    "Age": 20,
    "Marks": 90
}
print("After Adding New Student:", students)

# Delete one student
del students["Student5"]
print("After Deleting Student5:", students)

# Update marks
students["Student2"]["Marks"] = 82
print("After Updating Marks of Student2:", students)

# Print all students
print("\nFinal Student Details:")
for key, value in students.items():
    print(key, ":", value)
```

```python
# Quest 7:-Write a user-defined function calculate_area(length,width) Return the area. Also create another function calculate_perimeter(length,width).

def calculate_area(length, width):
    return length * width
def calculate_perimeter(length, width):
    return 2 * (length + width)
length = float(input("Enter length: "))
width = float(input("Enter width: "))

print("Area:", calculate_area(length, width))
print("Perimeter:", calculate_perimeter(length, width))
```

```python
# Quest 8:- Using Lambda Functions Perform ● Square of a number ● Cube of a number ● Find maximum of two numbers

square = lambda x: x**2
cube = lambda x: x**3
maximum = lambda x, y: max(x, y)

print(square(5))
print(cube(3))
print(maximum(10, 20))
```

```python
#Q14. Reshape, Transpose, Sort, Reverse & Split.

array = np.array([10, 20, 30, 40, 50, 60])

print("Original Array:", array)

# Reshape
reshaped = array.reshape(2, 3)
print("\nReshaped:")
print(reshaped)

# Transpose
print("\nTranspose:")
print(reshaped.T)

# Sort
unsorted = np.array([60, 20, 50, 10, 40, 30])
print("\nSorted:")
print(np.sort(unsorted))

# Reverse
print("\nReverse:")
print(array[::-1])

# Split into two arrays
split_arrays = np.split(array, 2)
print("\nSplit Arrays:")
print(split_arrays)
```

**Section E — Pandas & EDA**

```python
#import pandas:
import pandas as pd
```

```python
#quest 15:- Read the dataset. Display ● First 10 rows ● Last 10 rows ● Shape ● Columns ● Datatypes

df = pd.read_csv("../../datasets/external/students_50.csv")

print("First 10 Rows:")
display(df.head(10))

print("Last 10 Rows:")
display(df.tail(10))


print("Shape:")
print(df.shape)

print("\nColumns:")
print(df.columns)

print("\nDatatypes:")
print(df.dtypes)
```

```python
# Quest 16:- Question 16 Find ● Missing values ● Duplicate records Handle them appropriately.
#Step 1 — Find missing values
print("Missing Values:")
print(df.isnull().sum())

#Step 2 — Find duplicate records
print("Duplicate Records:")
print(df.duplicated().sum())

#Step 3 — Handle them
numeric_columns = df.select_dtypes(include=np.number).columns

for column in numeric_columns:
    df[column] = df[column].fillna(df[column].median())

df = df.drop_duplicates()

# checking again
print("Missing Values After Handling:")
print(df.isnull().sum())

print("\nDuplicates After Handling:")
print(df.duplicated().sum())
```

**Conclusion: There is no duplicate values or missing bcz this dateset i made by me only i have shown in step it there is missing/duplicated will be there then how it will be fill using fillna.**

```python
#Question 17 Generate ● Summary Statistics ● Mean ● Median ● Mode ● Standard Deviation

print("Summary Statistics:")
display(df.describe())

print("Mean:")
print(df[["age", "math_score", "reading_score", "writing_score"]].mean())

print("\nMedian:")
print(df[["age", "math_score", "reading_score", "writing_score"]].median())

print("\nMode:")
print(df[["age", "math_score", "reading_score", "writing_score"]].mode().iloc[0])

print("\nStandard Deviation:")
print(df[["age", "math_score", "reading_score", "writing_score"]].std())
```

```python
#Q18. Filtering, Sorting, GroupBy, Merge & Concatenate.

#1. Filtering ->> Let's find students who scored 80 or more in Maths.
filtered_df = df[df["math_score"] >= 80]

display(filtered_df)

#2. Sorting ->> Sort students according to their Maths score from highest to lowest.
sorted_df = df.sort_values(by="math_score", ascending=False)

display(sorted_df)

#3. GroupBy ->> Find the average Maths score according to gender.
grouped_df = df.groupby("gender")["math_score"].mean()

print(grouped_df)

#4. Merge ->> First create another DataFrame containing attendance.
attendance = pd.DataFrame({
    "student_id": df["student_id"],
    "attendance": np.random.randint(70, 101, len(df))
})

display(attendance.head())
merged_df = pd.merge(df, attendance, on="student_id")

display(merged_df.head(10))

#5. Concatenate ->>Split the dataset into two parts:
df1 = df.iloc[:25]
df2 = df.iloc[25:]
concatenated_df = pd.concat([df1, df2])

display(concatenated_df)
```

```python
#Question 19 Create a new column Example Total Marks Percentage Result Result should display Pass Fail .
df["Total Marks"] = df["math_score"] + df["reading_score"] + df["writing_score"]

df["Percentage"] = df["Total Marks"] / 3

df["Result"] = np.where(df["Percentage"] >= 40, "Pass", "Fail")

display(df.head(10))
```

# Section F: Data Visualization 

**Create the following charts. 1. Line Plot 2. Bar Plot 3. Scatter Plot 4. Histogram 5. Pie Chart 6. Boxplot 7. Heatmap For each graph, ● Write one observation. ● Mention what insights can be derived.**

```python
#import matplotlib and seaborn

import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# 1. Line Plot
plt.figure(figsize=(8, 5))

plt.plot(df["student_id"], df["math_score"], marker="o")

plt.title("Math Scores of Students")
plt.xlabel("Student ID")
plt.ylabel("Math Score")

plt.show()
```

Observation:
The Math scores are different for each student, with some students scoring much higher than others.

Insight:
This graph helps us see how the Math scores change from one student to another.

```python
#2. Bar Plot
gender_math = df.groupby("gender")["math_score"].mean()

plt.figure(figsize=(7, 5))

plt.bar(gender_math.index, gender_math.values)

plt.title("Average Math Score by Gender")
plt.xlabel("Gender")
plt.ylabel("Average Math Score")

plt.show()
```

Observation:
The average Math scores of male and female students are slightly different.

Insight:
We can easily compare the average Math performance of both groups using this graph.

```python
# 3. Scatter Plot
plt.figure(figsize=(8, 5))

plt.scatter(df["math_score"], df["reading_score"])

plt.title("Math Score vs Reading Score")
plt.xlabel("Math Score")
plt.ylabel("Reading Score")

plt.show()
```

Observation:
Students who have higher Math scores mostly have higher Reading scores as well.

Insight:
This shows that Math and Reading scores have a positive relationship.

```python
#4. Histogram
plt.figure(figsize=(8, 5))

plt.hist(df["math_score"], bins=10)

plt.title("Distribution of Math Scores")
plt.xlabel("Math Score")
plt.ylabel("Number of Students")

plt.show()
```

Observation:
Most students have Math scores in the middle score ranges, while fewer students have very low or very high scores.

Insight:
The graph helps us understand how the Math scores are spread among the students.

```python
# 5. pie Chart
gender_count = df["gender"].value_counts()

plt.figure(figsize=(7, 7))

plt.pie(
    gender_count.values,
    labels=gender_count.index,
    autopct="%1.1f%%"
)

plt.title("Gender Distribution")

plt.show()
```

Observation:
The dataset contains both male and female students, with one group having a slightly larger share.

Insight:
The pie chart gives us a quick idea of the gender distribution in the dataset.

```python
#6. Boxplot
plt.figure(figsize=(7, 5))

plt.boxplot(df["math_score"])

plt.title("Math Score Boxplot")
plt.ylabel("Math Score")

plt.show()
```

Observation:
Most Math scores are within a normal range, with a few scores being comparatively low or high.

Insight:
The boxplot helps us understand the spread of Math scores and identify unusual scores.

```python
#7. Heatmap
correlation = df[
    ["age", "math_score", "reading_score", "writing_score"]
].corr()

plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation,
    annot=True
)

plt.title("Correlation Heatmap")

plt.show()
```

Observation:
Math, Reading and Writing scores have a positive relationship with each other.

Insight:
The heatmap helps us see which subjects have a stronger relationship with each other.

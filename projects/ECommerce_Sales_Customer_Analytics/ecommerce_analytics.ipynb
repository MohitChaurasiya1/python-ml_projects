# Phase 2: File Handling & Data Storage

### 17. Save NumPy Data to CSV

```python
# Combine NumPy arrays into a single 2D array
sales_data = np.column_stack((
    sales_amount.flatten(),
    units_sold.flatten(),
    discount.flatten(),
    customer_rating.flatten()
))

print("Sales data shape:", sales_data.shape)
```

### 18. Save Data to CSV using `open()`

```python
# Define the CSV file path
file_path = "data/sales_data.csv"

# Open the file in write mode
with open(file_path, "w") as file:

    # Write the column headings
    file.write("Sales_Amount,Units_Sold,Discount,Customer_Rating\n")

    # Write each sales record
    for row in sales_data:
        file.write(
            f"{row[0]},{int(row[1])},{row[2]},{row[3]}\n"
        )

print("Sales data saved successfully!")
```

### 19. Store Metadata & Category Mapping in JSON

```python
import json

# Store metadata for each store
store_config = {
    "stores": {
        "Store A": {
            "location": "Lucknow",
            "category": "Electronics"
        },
        "Store B": {
            "location": "Delhi",
            "category": "Clothing"
        },
        "Store C": {
            "location": "Mumbai",
            "category": "Grocery"
        }
    },
    "categories": [
        "Electronics",
        "Clothing",
        "Grocery",
        "Furniture"
    ]
}

# Save store configuration to a JSON file
json_file_path = "data/store_config.json"

with open(json_file_path, "w") as file:
    json.dump(store_config, file, indent=4)

print("Store configuration saved successfully!")
```

### 20. Import CSV with Exception Handling

```python
# Import sales data from CSV with exception handling

file_path = "data/sales_data.csv"

try:
    with open(file_path, "r") as file:
        lines = file.readlines()

    sales_records = []

    for line in lines[1:]:
        try:
            values = line.strip().split(",")

            if len(values) != 4:
                continue

            sales = float(values[0])
            units = int(float(values[1]))
            discount_value = float(values[2])
            rating = float(values[3])

            sales_records.append(
                [sales, units, discount_value, rating]
            )

        except ValueError:
            print("Skipping corrupted record:", line.strip())
            continue

except FileNotFoundError:
    print("Error: CSV file not found.")

else:
    print("CSV imported successfully!")
    print("Number of records:", len(sales_records))

finally:
    print("CSV import process completed.")
```

### 21. Import JSON with Exception Handling

```python
# Import store configuration from JSON with exception handling

json_file_path = "data/store_config.json"

try:
    with open(json_file_path, "r") as file:
        config_data = json.load(file)

    stores_data = config_data["stores"]
    categories_data = config_data["categories"]

except FileNotFoundError:
    print("Error: JSON file not found.")

except KeyError as error:
    print("Error: Missing key in JSON file:", error)

else:
    print("JSON imported successfully!")
    print("Stores:", stores_data)
    print("Categories:", categories_data)

finally:
    print("JSON import process completed.")
```

### 22. Handle Corrupted and Out-of-Range Records

```python
# Validate sales records and skip invalid values

valid_records = []

for record in sales_records:
    sales = record[0]
    units = record[1]
    discount_value = record[2]
    rating = record[3]

    # Skip records with invalid sales or units
    if sales < 0 or units <= 0:
        continue

    # Skip records with invalid discount
    if discount_value < 0 or discount_value > 0.5:
        continue

    # Skip records with invalid customer rating
    if rating < 1 or rating > 5:
        continue

    valid_records.append(record)

print("Total records:", len(sales_records))
print("Valid records:", len(valid_records))
```

### 23. Control Flow with while, break and pass

```python
# Demonstrate while, break and pass during record validation

index = 0
checked_records = 0

while index < len(sales_records):
    record = sales_records[index]

    if record is None:
        pass

    checked_records += 1

    if checked_records >= 10:
        break

    index += 1

print("Records checked using while loop:", checked_records)
print("Control flow demonstration completed.")
```

# Phase 3: Pandas Data Processing & EDA

### 23. Load CSV Data into Pandas DataFrame

```python
# Load the sales CSV file into a Pandas DataFrame

sales_df = pd.read_csv("data/sales_data.csv")

print("Sales DataFrame:")
display(sales_df.head())

print("\nDataFrame Shape:", sales_df.shape)
```

### 24. Create DataFrames from Lists and Dictionaries

```python
# Create a DataFrame from lists

sample_data = [
    ["Store A", "Electronics", 5000],
    ["Store B", "Clothing", 3500],
    ["Store C", "Grocery", 4200]
]

sample_df = pd.DataFrame(
    sample_data,
    columns=["Store", "Category", "Sales"]
)

print("DataFrame from List:")
display(sample_df)


# Create a DataFrame from dictionary

store_df = pd.DataFrame(store_metadata).T

print("DataFrame from Dictionary:")
display(store_df)
```

### 25. Insert and Delete Columns

```python
# Insert a new column for Net Profit

sales_df.insert(
    4,
    "Net_Profit",
    sales_df["Sales_Amount"] * (1 - sales_df["Discount"])
)

print("DataFrame after inserting Net Profit:")
display(sales_df.head())


# Delete the Net Profit column

sales_df.drop("Net_Profit", axis=1, inplace=True)

print("DataFrame after deleting Net Profit:")
display(sales_df.head())
```

### 26. Calculate Net Profit

```python
# Calculate Net Profit using Sales Amount and Discount

sales_df["Net_Profit"] = (
    sales_df["Sales_Amount"] * (1 - sales_df["Discount"])
)

print("DataFrame with Net Profit:")
display(sales_df.head())
```

### 27. Inject and Check Missing Values

```python
# Create a copy before injecting missing values
sales_df = sales_df.copy()

# Randomly select 5% of the rows
missing_rows = sales_df.sample(
    frac=0.05,
    random_state=42
).index

# Replace Sales Amount with NaN for selected rows
sales_df.loc[missing_rows, "Sales_Amount"] = np.nan

# Check the number of missing values
print("Missing values in each column:")
print(sales_df.isnull().sum())
```

### 28. Handle Missing Values using fillna()

```python
# Fill missing Sales Amount values with the mean sales amount

sales_mean = sales_df["Sales_Amount"].mean()

sales_df["Sales_Amount"] = sales_df["Sales_Amount"].fillna(sales_mean)

# Recalculate Net Profit after handling missing Sales Amount values
sales_df["Net_Profit"] = (
    sales_df["Sales_Amount"] * (1 - sales_df["Discount"])
)

print("Missing values after fillna():")
print(sales_df.isnull().sum())
```

### 29. Detect and Remove Duplicate Records

```python
# Check the number of duplicate records

duplicate_count = sales_df.duplicated().sum()

print("Number of duplicate records:", duplicate_count)


# Remove duplicate records

sales_df = sales_df.drop_duplicates()

print("DataFrame shape after removing duplicates:", sales_df.shape)
```

### 30. Detect and Treat Sales Outliers using IQR

```python
# Calculate Q1 and Q3
Q1 = sales_df["Sales_Amount"].quantile(0.25)
Q3 = sales_df["Sales_Amount"].quantile(0.75)

# Calculate IQR
IQR = Q3 - Q1

# Define lower and upper limits
lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

# Detect outliers
outliers = sales_df[
    (sales_df["Sales_Amount"] < lower_limit) |
    (sales_df["Sales_Amount"] > upper_limit)
]

print("Number of Sales Outliers:", len(outliers))


# Treat outliers by replacing them with the median
sales_median = sales_df["Sales_Amount"].median()

sales_df.loc[
    sales_df["Sales_Amount"] < lower_limit,
    "Sales_Amount"
] = sales_median

sales_df.loc[
    sales_df["Sales_Amount"] > upper_limit,
    "Sales_Amount"
] = sales_median

# Recalculate Net Profit
sales_df["Net_Profit"] = (
    sales_df["Sales_Amount"] * (1 - sales_df["Discount"])
)

print("Outliers treated successfully.")
```

### 31. Create Date Column and Convert to DatetimeIndex

```python
# Create dates for 365 days and repeat them for 3 stores
dates = pd.date_range(
    start="2025-01-01",
    periods=days,
    freq="D"
)

sales_df["Date"] = np.tile(dates, stores)

# Convert Date column to datetime
sales_df["Date"] = pd.to_datetime(sales_df["Date"])

# Set Date as the DataFrame index
sales_df = sales_df.set_index("Date")

print("DataFrame index type:", type(sales_df.index))
print("\nFirst 5 records:")
display(sales_df.head())
```

### 32. Extract Month, Day and Day of Week

```python
# Extract month, day and day of week from the DatetimeIndex

sales_df["Month"] = sales_df.index.month
sales_df["Day"] = sales_df.index.day
sales_df["DayOfWeek"] = sales_df.index.day_name()

print("Date-based columns added successfully:")
display(sales_df.head())
```

### 33. Time Resampling


```python
# Resample daily sales data into monthly sales totals

monthly_sales_summary = sales_df["Sales_Amount"].resample("ME").sum()

print("Monthly Sales Summary:")
print(monthly_sales_summary)
```

### 34. Merge Sales Data with Store Metadata

```python
# Reset index temporarily so Date becomes a column
sales_merge_df = sales_df.reset_index()

# Create a DataFrame from store metadata
store_metadata_df = pd.DataFrame.from_dict(
    store_metadata,
    orient="index"
).reset_index()

# Rename columns for merging
store_metadata_df.columns = ["Store", "Location", "Store_Category"]

# Add store names to sales data
sales_merge_df["Store"] = np.tile(
    ["Store A", "Store B", "Store C"],
    days
)

# Merge sales data with store metadata
merged_df = pd.merge(
    sales_merge_df,
    store_metadata_df,
    on="Store",
    how="left"
)

print("Merged DataFrame:")
display(merged_df.head())
```

### 35. Concatenate DataFrames

```python
# Create two smaller DataFrames from the merged data

df_part1 = merged_df.iloc[:500]
df_part2 = merged_df.iloc[500:]

# Concatenate the two DataFrames
concatenated_df = pd.concat(
    [df_part1, df_part2],
    axis=0
)

print("Concatenated DataFrame shape:", concatenated_df.shape)
display(concatenated_df.head())
```

### 36. Join DataFrames

```python
# Create sales data with Store as the index
sales_join_df = merged_df.set_index("Store")

# Create store metadata DataFrame
store_join_df = pd.DataFrame.from_dict(
    store_metadata,
    orient="index"
)

# Join store metadata with sales data
joined_df = sales_join_df.join(
    store_join_df,
    rsuffix="_metadata"
)

print("Joined DataFrame:")
display(joined_df.head())
```

### 37. GroupBy Analysis

```python
# Calculate average sales and units sold by store

store_analysis = merged_df.groupby("Store")[
    ["Sales_Amount", "Units_Sold"]
].mean()

print("Average Sales and Units by Store:")
display(store_analysis)


# Calculate average sales and units sold by category

category_analysis = merged_df.groupby("Store_Category")[
    ["Sales_Amount", "Units_Sold"]
].mean()

print("Average Sales and Units by Category:")
display(category_analysis)


# Calculate average sales and units sold by month

month_analysis = merged_df.groupby("Month")[
    ["Sales_Amount", "Units_Sold"]
].mean()

print("Average Sales and Units by Month:")
display(month_analysis)
```

### 38. Final DataFrame Check

```python
# Display the first five records
display(sales_df.head())

# Check DataFrame shape
print("DataFrame Shape:", sales_df.shape)

# Check data types
print("\nData Types:")
print(sales_df.dtypes)

# Check remaining missing values
print("\nMissing Values:")
print(sales_df.isnull().sum())
```
